```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Z-Score Normalization
np.random.seed(42)
X = np.array([
    [25, 50000, 85],
    [32, 64000, 92],
    [47, 120000, 78],
    [19, 22000, 88],
    [55, 95000, 95]
], dtype=np.float64)

X_scaled = (X - np.mean(X, axis=0)) / np.std(X, axis=0)
print("Standardized Features Matrix:\n", np.round(X_scaled, 2))

# 2. Create a NumPy 3D Image Matrix
img = np.zeros((100, 100, 3), dtype=np.uint8)
img[:50, :50] = [255, 0, 0]    # Red quadrant
img[50:, 50:] = [0, 255, 0]    # Green quadrant
img[:50, 50:] = [0, 0, 255]    # Blue quadrant
img[50:, :50] = [255, 255, 0]  # Yellow quadrant

plt.imshow(img)
plt.axis('off')
plt.show()
```
